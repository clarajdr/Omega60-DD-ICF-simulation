```python
import pandas as pd
import matplotlib.pyplot as plt

# --- 1. CONFIGURATION DES FICHIERS ---
fichier_ref = 'Analyse_+15energy.xlsx'
fichier_res = 'Analyse_Resol256.xlsx'
nom_onglet = 'Simulation Evolution'

print("Lecture des fichiers Excel en cours...")
# Lecture directe de l'onglet contenant les données brutes
df_ref = pd.read_excel(fichier_ref, sheet_name=nom_onglet)
df_res = pd.read_excel(fichier_res, sheet_name=nom_onglet)

# --- 2. GÉNÉRATION DES GRAPHIQUES ---
print("Génération des graphiques de comparaison...")
label_ref = "Référence (512x512)"
label_res = "Résolx2 (1024x1024)"

fig, axs = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Comparaison des Simulations ARWEN (depuis Excel)', fontsize=16, fontweight='bold')

# 1. Comparaison Pression Maximale
axs[0, 0].plot(df_ref['Time (s)'], df_ref['Max Pressure (Pa)'], label=label_ref, color='blue', linestyle='--')
axs[0, 0].plot(df_res['Time (s)'], df_res['Max Pressure (Pa)'], label=label_res, color='red', alpha=0.7)
axs[0, 0].set_title("Pression Maximale du maillage")
axs[0, 0].set_xlabel("Temps (s)")
axs[0, 0].set_ylabel("Pression (Pa)")
axs[0, 0].grid(True)
axs[0, 0].legend()

# 2. Comparaison Température Maximale
axs[0, 1].plot(df_ref['Time (s)'], df_ref['Max Temp (eV)'], label=label_ref, color='blue', linestyle='--')
axs[0, 1].plot(df_res['Time (s)'], df_res['Max Temp (eV)'], label=label_res, color='red', alpha=0.7)
axs[0, 1].set_title("Température Maximale")
axs[0, 1].set_xlabel("Temps (s)")
axs[0, 1].set_ylabel("Température (eV)")
axs[0, 1].grid(True)
axs[0, 1].legend()

# 3. Comparaison Densité Matériau 1 (DT)
axs[1, 0].plot(df_ref['Time (s)'], df_ref['Density D1 (kg/m3)'], label=label_ref, color='blue', linestyle='--')
axs[1, 0].plot(df_res['Time (s)'], df_res['Density D1 (kg/m3)'], label=label_res, color='red', alpha=0.7)
axs[1, 0].set_title("Évolution Densité Matériau 1 (DT)")
axs[1, 0].set_xlabel("Temps (s)")
axs[1, 0].set_ylabel("Densité (kg/m3)")
axs[1, 0].grid(True)
axs[1, 0].legend()

# 4. Comparaison Densité Matériau 2 (CH)
axs[1, 1].plot(df_ref['Time (s)'], df_ref['Density D2 (kg/m3)'], label=label_ref, color='blue', linestyle='--')
axs[1, 1].plot(df_res['Time (s)'], df_res['Density D2 (kg/m3)'], label=label_res, color='red', alpha=0.7)
axs[1, 1].set_title("Évolution Densité Matériau 2 (CH)")
axs[1, 1].set_xlabel("Temps (s)")
axs[1, 1].set_ylabel("Densité (kg/m3)")
axs[1, 1].grid(True)
axs[1, 1].legend()

plt.tight_layout(rect=[0, 0.03, 1, 0.95])

# --- 3. SAUVEGARDE DE L'IMAGE ---
plot_filename = 'Comparaison_Ref_vs_Resol2.png'
plt.savefig(plot_filename, dpi=300)
print(f"Terminé ! Graphique sauvegardé sous : {plot_filename}")
```
